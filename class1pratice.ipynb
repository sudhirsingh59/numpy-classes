```python
import numpy as np
```

```python
#Q1.  1 se 100 tak numbers banao (arange)
arr = np.arange(1, 100)
print(arr)
```

```python
# 1 to 100 tak numbers banao phir us number ka mean, max, min or sum nikalo.
arr = np.arange(1,100)
print("Mean",np.mean(arr))
print("Max",np.max(arr))
print("Min",np.min(arr))
print("Sum",np.sum(arr))
```

```python
# Q2. 0 se 1 kai each 50 equal space banwo(linspace)
arr = np.linspace(0,1,50)
print(arr)
```

```python
#Dono ka:
#Mean
#Max
#Min
#Sum nikaalo
arr = np.linspace(0,1,50)
print("Mean",np.mean(arr))
print("Max",np.max(arr))
print("Min",np.min(arr))
print("Sum",np.sum(arr))
```

```python
# 5×5 identity matrix banao (eye)
arr = np.eye(5)
print(arr)
```

```python
# random
arr = np.random.rand(2,3)
print(arr)
```

```python
# A × I = A
# I × A = A
A = np.random.randint(1,10,(5,5))
I = np.eye(5,dtype=int)

print("Matrix A:\n",A)
print("\n Identity Matrix I: \n",I)

print("\nA * I =\n",A@I)
print("\nI * A = \n",I @ A)
```
